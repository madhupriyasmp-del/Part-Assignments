```python
import seaborn as sns
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")
df
```

```python
#Check for missing values in the dataset and identify columns with missing data.
print(df.isna().sum())
missing_colums = df.columns[df.isna().any()]
print(missing_colums)
```

```python
#Impute missing values using appropriate strategies based on the column type (e.g., using the mean, median, or mode for numerical columns and the mode for categorical columns

df["passengers"] = df["passengers"].fillna(df["passengers"].median())
df["distance"] = df["distance"].fillna(df["distance"].median())
df["fare"] = df["fare"].fillna(df["fare"].median())
df["tip"] = df["tip"].fillna(df["tip"].median())
df["tolls"] = df["tolls"].fillna(df["tolls"].median())
df["total"] = df["total"].fillna(df["total"].median())

df["color"] = df["color"].fillna(df["color"].mode()[0])
df["payment"] = df["payment"].fillna(df["payment"].mode()[0])
df["pickup_zone"] = df["pickup_zone"].fillna(df["pickup_zone"].mode()[0])
df["dropoff_zone"] = df["dropoff_zone"].fillna(df["dropoff_zone"].mode()[0])
df["pickup_borough"] = df["pickup_borough"].fillna(df["pickup_borough"].mode()[0])
df["dropoff_borough"] = df["dropoff_borough"].fillna(df["dropoff_borough"].mode()[0])

df
```

```python
#For columns that are critical and cannot be reasonably imputed, remove rows with missing values to maintain data integrity.
df = df.dropna(subset=["pickup", "dropoff"])

print(df.isna().sum())
```

```python
#★ Line Chart
#Plot a line chart to visualize the fare over time, using the pickup timestamp as the x-axis and fare as the y-axis. Ensure the pickup column is converted to a datetime format before plotting.


import pandas as pd
import matplotlib.pyplot as plt


df["pickup"] = pd.to_datetime(df["pickup"])
df = df.sort_values("pickup")

plt.figure(figsize=(10, 5))
plt.plot(df["pickup"], df["fare"])


plt.title("Fare Over Time")
plt.xlabel("Pickup Timestamp")
plt.ylabel("Fare Amount")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#★ Bar Chart
#Create a bar chart to show the total fare for each pickup_borough. Group the data by pickup_borough and sum the fare for each group.


borough_fare = df.groupby("pickup_borough")["fare"].sum().reset_index()

plt.figure(figsize=(10,5))
plt.bar(borough_fare["pickup_borough"],
        borough_fare["fare"],
        color='green',
        edgecolor='red',
        linewidth=2,
        align='center',
        alpha = 0.8
        )
plt.title("Total Fare by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#★ Pie Chart
#Plot a pie chart showing the distribution of trips based on the payment method (credit card, cash, etc.). Each slice should represent the count of trips for a specific payment method.


payment_counts = df["payment"].value_counts()


plt.pie(payment_counts.values,
        labels=payment_counts.index,
        autopct="%1.1f%%",
        startangle=90,
        colors=["green", "orange", "skyblue", "pink", "yellow"],
        #explode=[0.05, 0, 0, 0, 0]
        shadow=True
        )

plt.title("Distribution of Trips by Payment Method")
plt.show()
```

```python
#★ Histogram
#Create a histogram to visualize the distribution of distance. Customize the number of bins for better granularity and ensure the plot is easy to interpret.

plt.figure(figsize=(8, 5))

plt.hist(df["distance"].dropna(),
         bins=30, color="skyblue",
         edgecolor="black")

plt.title("Distribution of Taxi Trip Distances")
plt.xlabel("Distance")
plt.ylabel("Number of Trips")

plt.tight_layout()
plt.show()
```

```python
#★ Box Plot
#Plot a box plot to visualize the distribution of tip amounts for each pickup_borough. Use pickup_borough as the categorical axis and tip as the numeric axis.

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="pickup_borough", y="tip", color="skyblue")

plt.title("Distribution of Tip Amounts by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
#★ Count Plot
#Create a count plot to visualize the number of trips in each pickup_borough. The x-axis should represent the boroughs, and the y-axis should show the count of trips.
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="pickup_borough", color="skyblue")

plt.title("Number of Trips by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Count of Trips")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
#★ Scatter Plot
#Plot a scatter plot to show the relationship between distance and fare. Use distance on the x-axis and fare on the y-axis to visualize any correlation. Color the points based on the
#pickup_borough to differentiate the trips by their respective boroughs.


plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="distance",
    y="fare",
    hue="pickup_borough"
)

plt.title("Relationship Between Distance and Fare by Pickup Borough")
plt.xlabel("Distance")
plt.ylabel("Fare Amount")
plt.legend(bbox_to_anchor=(1,1))
plt.tight_layout()
plt.show()
```

```python
#★ Heatmap
#Plot a heatmap to visualize the correlation between numerical variables such as distance, fare, tip, tolls, and total. Use a correlation matrix to highlight the relationships.

numerical_data = df[["distance", "fare", "tip", "tolls", "total"]]

correlation_matrix = numerical_data.corr()

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,annot=True,
    )

plt.title("Correlation Heatmap of Taxi Data")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
#★ Pair Plot
#Create a pair plot to visualize the pairwise relationships between distance, fare, tip, and total. Color the data points according to the pickup_zone column. to compare how different zones
#affect these variables.

pair_data = df[["distance", "fare", "tip", "total", "pickup_zone"]].dropna()

sns.pairplot(
    data=pair_data,
    vars=["distance", "fare", "tip", "total"],
    hue="pickup_zone",
    diag_kind="hist"
)

plt.show()
```

```python
#★ Violin Plot
#Plot a violin plot to show the distribution of fare for each payment method. Use the payment method as the categorical axis and fare as the numeric axis to visualize its distribution.



plt.figure(figsize=(10, 5))

sns.violinplot(
    data=df,
    x="payment",
    y="fare",
    color="skyblue"
)

plt.title("Distribution of Fare by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Fare Amount")

plt.tight_layout()
plt.show()
```
