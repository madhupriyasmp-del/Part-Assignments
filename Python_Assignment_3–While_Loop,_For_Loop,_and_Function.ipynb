```python
## Task 1: While Loop & Control Statements -Number Guessing Game

import random
number = random.randint(1,10)
print(number)

attempts = 0
while attempts<3:

    guess = int(input("Guess the number (between 1 and 10): "))
    attempts+=1

    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please enter a number between 1 and 10.")
        continue
    if guess == number:
        print("Congratulations! You guessed the correct number.")
        break
    elif guess>number:
        print("Too high. Try again.")
    else:
        print("Too low. Try again.")
if attempts == 3 and guess != number:
    print("You have reached your maximum attempts.")
```

```python
## Task 2:- For Loop - Multiplication Table Generator

table = int(input("Enter the number for which you want the multiplication table: "))

for i in range(1, 11):
    print(f"{table} * {i} = {table * i}")
```

```python
## Task 3: Function- BMI Calculator using return

def calculate_bmi(weight, height):
    return weight / (height ** 2)


weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

BMI = calculate_bmi(weight, height)

print("Your BMI is:", BMI)
```
